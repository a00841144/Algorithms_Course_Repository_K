```python
!pip install yfinance --quiet
import yfinance as yf
import pandas as pd
import numpy as np

tickers = ["WALMEX.MX", "AC.MX", "LIVEPOLC.MX", "BIMBOA.MX", "ALSEA.MX", "AAPL", "MSFT", "AMZN", "KO", "PG"]

raw_data = yf.download(tickers, start="2021-08-31", end="2026-08-31", group_by='ticker', auto_adjust=True, progress=False)

frames = []
for t in tickers:
    df_t = raw_data[t].copy()
    df_t['Ticker'] = t
    df_t.reset_index(inplace=True)
    frames.append(df_t)

fondo = pd.concat(frames, ignore_index=True)
fondo = fondo[['Date', 'Ticker', 'Open', 'High', 'Low', 'Close', 'Volume']]
fondo.sort_values(['Ticker', 'Date'], inplace=True)
fondo = fondo.dropna(subset=['Close']).drop_duplicates(subset=['Date','Ticker'])
fondo.reset_index(drop=True, inplace=True)
fondo.head()
```
