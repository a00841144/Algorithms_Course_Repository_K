## Question 2 Check for missing values and replace them with the mean.

In this part, we first check how many missing values we have in each column.

```python
mis_datos.isna().sum()
```

As we can see, we have several missing values, in EngineSize, Cylinders, Horsepower, MPG_City       2
MPG_Highway    2
Weight         5
Wheelbase      2
Length         1

```python
# EngineSize
average_enginesize = mis_datos["EngineSize"].mean()
mis_datos["EngineSize"] = (
    mis_datos["EngineSize"]
    .fillna(average_enginesize)
)

# Cylinders
average_cylinders = mis_datos["Cylinders"].mean()
mis_datos["Cylinders"] = (
    mis_datos["Cylinders"]
    .fillna(average_cylinders)
)

# Horsepower
average_horsepower = mis_datos["Horsepower"].mean()
mis_datos["Horsepower"] = (
    mis_datos["Horsepower"]
    .fillna(average_horsepower)
)

# MPG_City
average_mpg_city = mis_datos["MPG_City"].mean()
mis_datos["MPG_City"] = (
    mis_datos["MPG_City"]
    .fillna(average_mpg_city)
)

# MPG_Highway
average_mpg_highway = mis_datos["MPG_Highway"].mean()
mis_datos["MPG_Highway"] = (
    mis_datos["MPG_Highway"]
    .fillna(average_mpg_highway)
)

# Weight
average_weight = mis_datos["Weight"].mean()
mis_datos["Weight"] = (
    mis_datos["Weight"]
    .fillna(average_weight)
)

# Wheelbase
average_wheelbase = mis_datos["Wheelbase"].mean()
mis_datos["Wheelbase"] = (
    mis_datos["Wheelbase"]
    .fillna(average_wheelbase)
)

# Length
average_length = mis_datos["Length"].mean()
mis_datos["Length"] = (
    mis_datos["Length"]
    .fillna(average_length)
)
```
