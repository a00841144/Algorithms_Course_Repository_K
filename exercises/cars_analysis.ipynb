## Question 2 Check for missing values and replace them with the mean.

In this part, we first check how many missing values we have in each column.

```python
mis_datos.isna().sum()
```

As we can see, we have several missing values, in EngineSize, Cylinders, Horsepower, MPG_City, MPG_Highway, Weight, Wheelbase, Length.
We will fill the missing values with the mean of each column.

```python
# EngineSize
average_enginesize = mis_datos["EngineSize"].mean()
mis_datos["EngineSize"] = (
    mis_datos["EngineSize"]
    .fillna(average_enginesize)
)

# Cylinders
average_cylinders = mis_datos["Cylinders"].mean()
mis_datos["Cylinders"] = (
    mis_datos["Cylinders"]
    .fillna(average_cylinders)
)

# Horsepower
average_horsepower = mis_datos["Horsepower"].mean()
mis_datos["Horsepower"] = (
    mis_datos["Horsepower"]
    .fillna(average_horsepower)
)

# MPG_City
average_mpg_city = mis_datos["MPG_City"].mean()
mis_datos["MPG_City"] = (
    mis_datos["MPG_City"]
    .fillna(average_mpg_city)
)

# MPG_Highway
average_mpg_highway = mis_datos["MPG_Highway"].mean()
mis_datos["MPG_Highway"] = (
    mis_datos["MPG_Highway"]
    .fillna(average_mpg_highway)
)

# Weight
average_weight = mis_datos["Weight"].mean()
mis_datos["Weight"] = (
    mis_datos["Weight"]
    .fillna(average_weight)
)

# Wheelbase
average_wheelbase = mis_datos["Wheelbase"].mean()
mis_datos["Wheelbase"] = (
    mis_datos["Wheelbase"]
    .fillna(average_wheelbase)
)

# Length
average_length = mis_datos["Length"].mean()
mis_datos["Length"] = (
    mis_datos["Length"]
    .fillna(average_length)
)
```

```python
mis_datos.isna().sum()
```

After re-checking, now we have no missing values.

### Question 5. The highest suggested retail price, the lowest suggested retail price, the average city miles per gallon, and the average highway miles per gallon.

```python
highest_sprice = mis_datos["MSRP"].max()
lowest_sprice = mis_datos["MSRP"].min()

highest_car = mis_datos.loc[mis_datos["MSRP"].idxmax()]
lowest_car = mis_datos.loc[mis_datos["MSRP"].idxmin()]

average_mpg_city = mis_datos["MPG_City"].mean()
average_mpg_highway = mis_datos["MPG_Highway"].mean()

print(f"Highest MSRP: ${highest_sprice:,.2f} - {highest_car['Make']} {highest_car['Model']}")
print(f"Lowest MSRP: ${lowest_sprice:,.2f} - {lowest_car['Make']} {lowest_car['Model']}")

print(f"Average MPG (city): {average_mpg_city:.2f}")
print(f"Average MPG (highway): {average_mpg_highway:.2f}")
```

As we can see, the highest suggested retail price of $192,465.00,  corresponds to the Porsche 911 GT2 2dr. And the lowest to a Kia Rio 4dr manual of $10,280.00.
Another insight is that cars on average perform better on the highway (26.84 mpg), than on the city (20.07).

### Question 6. Select only the following variables: Make, Model, Type, Origin, DriveTrain, MSRP, Weight, Wheelbase, and Length. Keep only the observations where DriveTrain is equal to Front, and store this information in a variable called mis_nuevos_datos.

### Question 7
Print the first observations of the mis_nuevos_datos variable.
Using the information stored in mis_nuevos_datos, create three new variables:
• Weight_kilo, which indicates the weight of the vehicle in kilograms.
• Wheelbase_cm, which shows the value of the Wheelbase variable in centimeters.
• Length_cm, which indicates the length of the vehicle in centimeters.
Store all this information in the mis_nuevos_datos variable.

```python
mis_nuevos_datos.head()
```

Here we can see a view of the first rows of the new data set.

```python
LB_TO_KG = 0.453592
IN_TO_CM = 2.54

mis_nuevos_datos["Weight_kilo"] = mis_nuevos_datos["Weight"] * LB_TO_KG
mis_nuevos_datos["Wheelbase_cm"] = mis_nuevos_datos["Wheelbase"] * IN_TO_CM
mis_nuevos_datos["Length_cm"] = mis_nuevos_datos["Length"] * IN_TO_CM

mis_nuevos_datos.head()
```

I created three new columns, reflecting new data derived from the original data set, which is useful as the typical measurements units we use in México are KG and Cm, not Lb and IN. However, the orihinal data is still there for comparison.

### Question 8
Using the information stored in mis_nuevos_datos, create a new variable called opinion
based on the suggested retail price.
• If the price is greater than 60,000, the value of opinion must be "Expensive".
• If the price is between 30,000 and 60,000, the value of opinion must be "Normal".
• If the price is less than 30,000, the value of opinion must be "Inexpensive".
Store the new variable in mis_nuevos_datos.

```python
def opinion_p(price):
    if price > 60000:
        return "Expensive"
    elif price >= 30000:
        return "Normal"
    else:
        return "Inexpensive"

mis_nuevos_datos["opinion"] = mis_nuevos_datos["MSRP"].apply(opinion_p)

mis_nuevos_datos[["Make", "Model", "MSRP", "opinion"]].sample(10)
```

The new opinion column, now allow us to classify the price of each car, in order to facilitate the analysis. Here you can see a random sample of selected columns of the new data set.

```python
mis_nuevos_datos["opinion"].value_counts()
```

We can also get the insight that there are 175 inexpensive cars, 48 normal, and 3 expensive.
